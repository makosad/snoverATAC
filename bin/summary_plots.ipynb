```python

import sys
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# summary_path = sys.argv[1]
summary_path = '/home/dmakosa/working_data_02/snoverATAC/final.summary.csv'
# Read in the dependencies
summary = pd.read_csv(summary_path, sep = ',')
```

```python
summary
```

```python
summary = summary.loc[~summary['Sample ID'].str.contains('combined|bulk', case=False)]
```

```python
summary
```

```python
RL3210 = summary.loc[summary['Sample ID'].str.contains('RL3210', case=False)]
RL3294 = summary.loc[summary['Sample ID'].str.contains('RL3294', case=False)]
RL3295 = summary.loc[summary['Sample ID'].str.contains('RL3295', case=False)]
RL4632 = summary.loc[summary['Sample ID'].str.contains('RL4632', case=False)]
```

```python
RL3210[['RL','well','cleanup','TagBuffer','TagTemp','MErev','Tn5conc']] = RL3210['Sample ID'].str.split("_", expand = True)
RL3294[['RL','well','cleanup','TagBuffer','TagTemp','MErev','Tn5conc']] = RL3294['Sample ID'].str.split("_", expand = True)
RL3295[['RL','well','cleanup','TagBuffer','TagTemp','MErev','Tn5conc']] = RL3295['Sample ID'].str.split("_", expand = True)
```

```python
RL4632[['RL','well','Specie','SampleSource','LysisBuffer','TagBuffer']] = RL4632['Sample ID'].str.split("_", expand = True)
```

```python
RL4632
```

```python
final = pd.concat([RL3210, RL3294, RL3295, RL4632], ignore_index=True, sort=False)
```

```python
final
```

```python
sns.set_style("ticks")
sns.boxplot(data = final, x = "RL", y = 'Fraction of high-quality fragments in cells')
sns.stripplot(data = final, x = "RL", y = 'Fraction of high-quality fragments in cells', size=4, color=".3", linewidth=0)
plt.xticks(rotation=60, horizontalalignment='right')
plt.tight_layout()
```
